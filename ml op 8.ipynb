```python
# ✅ MP Neuron Implementation of Logic Gates

import numpy as np

# -----------------------------
# ✅ MP Neuron function
# -----------------------------
def mp_neuron(inputs, weights, threshold):
    total = np.dot(inputs, weights)
    return 1 if total >= threshold else 0


# -----------------------------
# ✅ AND Gate using MP Neuron
# -----------------------------
print("✅ AND Gate:")
weights = [1, 1]
threshold = 2

for x1 in [0, 1]:
    for x2 in [0, 1]:
        out = mp_neuron([x1, x2], weights, threshold)
        print(f"{x1} AND {x2} = {out}")

# -----------------------------
# ✅ OR Gate using MP Neuron
# -----------------------------
print("\n✅ OR Gate:")
weights = [1, 1]
threshold = 1

for x1 in [0, 1]:
    for x2 in [0, 1]:
        out = mp_neuron([x1, x2], weights, threshold)
        print(f"{x1} OR {x2} = {out}")

# -----------------------------
# ✅ NOT Gate using MP Neuron
# -----------------------------
print("\n✅ NOT Gate:")
weights = [-1]     # negative weight
threshold = -0.5

for x in [0, 1]:
    out = mp_neuron([x], weights, threshold)
    print(f"NOT {x} = {out}")
```
