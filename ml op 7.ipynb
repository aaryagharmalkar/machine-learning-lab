```python
# ✅ Gradient Descent for House Price Prediction
# ✅ Using REAL dataset: California Housing (sklearn)

from sklearn.datasets import fetch_california_housing
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# ✅ Step 1: Load real house price dataset
# ---------------------------------------------------------
data = fetch_california_housing()
X = data.data[:, 0]      # Feature: Median Income
y = data.target          # Target: House Price

# Optional: Reduce size for simplicity
X = X[:500]
y = y[:500]

# ---------------------------------------------------------
# ✅ Step 2: Initialize parameters
# ---------------------------------------------------------
m = 0   # slope
c = 0   # intercept
lr = 0.01
epochs = 300
loss_list = []
n = len(X)

# ---------------------------------------------------------
# ✅ Step 3: Gradient Descent
# ---------------------------------------------------------
for i in range(epochs):
    y_pred = m*X + c

    loss = (1/n) * np.sum((y_pred - y)**2)
    loss_list.append(loss)

    dm = (2/n) * np.sum((y_pred - y) * X)
    dc = (2/n) * np.sum(y_pred - y)

    m -= lr * dm
    c -= lr * dc

print("✅ Slope (m):", m)
print("✅ Intercept (c):", c)
print("✅ Final Loss:", loss_list[-1])

# ---------------------------------------------------------
# ✅ Step 4: Loss Curve
# ---------------------------------------------------------
plt.plot(loss_list)
plt.title("Loss Curve (Gradient Descent)")
plt.xlabel("Epochs")
plt.ylabel("MSE Loss")
plt.show()

# ---------------------------------------------------------
# ✅ Step 5: Regression Line on Real Dataset
# ---------------------------------------------------------
plt.scatter(X, y, color="blue", label="Actual Data", alpha=0.5)
plt.plot(X, m*X + c, color="red", label="Predicted Line")
plt.title("California Housing Prices (1 Feature)")
plt.xlabel("Median Income")
plt.ylabel("House Price")
plt.legend()

plt.show()

# ---------------------------------------------------------
# ✅ Step 6: Prediction
# ---------------------------------------------------------
test_income = 5.0    # income = 5 (50K)
pred_price = m*test_income + c

print("✅ Predicted house price for income=5.0:", pred_price)
```
