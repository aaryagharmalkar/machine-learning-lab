```python
import numpy as np
from sklearn.datasets import fetch_california_housing
from sklearn.preprocessing import StandardScaler

# --------------------------
# 1. LOAD CALIFORNIA DATASET
# --------------------------
housing = fetch_california_housing()
X = housing.data[:, 0:1]   # Use only "MedInc" as single feature
y = housing.target         # House price

# --------------------------
# 2. NORMALIZE FEATURE
# --------------------------
scaler = StandardScaler()
X = scaler.fit_transform(X)

# --------------------------
# 3. INITIALIZE PARAMETERS
# --------------------------
m = 0.0     # slope
b = 0.0     # intercept
lr = 0.01
epochs = 500

n = len(X)

# --------------------------
# 4. GRADIENT DESCENT LOOP
# --------------------------
for i in range(epochs):

    # Predictions
    y_pred = m * X + b

    # Compute gradients
    dm = (-2/n) * np.sum((y - y_pred) * X)
    db = (-2/n) * np.sum(y - y_pred)

    # Update parameters
    m -= lr * dm
    b -= lr * db

    if i % 50 == 0:
        loss = np.mean((y - y_pred)**2)
        print(f"Epoch {i}, Loss = {loss:.4f}")

# --------------------------
# 5. RESULTS
# --------------------------
print("\nFinal slope (m):", m)
print("Final intercept (b):", b)

# --------------------------
# 6. TEST PREDICTION
# --------------------------
test_income = 5.0   # Suppose median income = 5.0 (scaled)
test_income_scaled = scaler.transform([[test_income]])
prediction = m * test_income_scaled + b

print("\nPredicted house price:", prediction[0])
```
