```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# 1. Simple Dataset (AND Gate)
# -----------------------------
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([0, 0, 0, 1])   # AND output

# -----------------------------
# 2. Initialize parameters
# -----------------------------
weights = np.zeros(2)
bias = 0
lr = 0.1   # learning rate

print("Initial Weights:", weights)
print("Initial Bias:", bias)

# For plotting weight change
w_history = []

# -----------------------------
# 3. Hebbian Learning Rule
# -----------------------------
for epoch in range(5):   # run few epochs only
    print(f"\nEpoch {epoch+1}")

    for i in range(len(X)):
        x_i = X[i]
        y_i = y[i]

        # Δw = η * x * y
        weights += lr * x_i * y_i
        bias += lr * y_i

        print(f"Input: {x_i}, Target: {y_i}, New Weights: {weights}")

    w_history.append(weights.copy())

# -----------------------------
# 4. Final weights
# -----------------------------
print("\nFinal learned weights:", weights)
print("Final learned bias:", bias)

# -----------------------------
# 5. Visualization of learning
# -----------------------------
w_history = np.array(w_history)

plt.plot(w_history[:, 0], label="Weight 1")
plt.plot(w_history[:, 1], label="Weight 2")
plt.xlabel("Epoch")
plt.ylabel("Weight Value")
plt.title("Hebbian Learning – Weight Update Over Epochs")
plt.legend()
plt.grid()
plt.show()
```
